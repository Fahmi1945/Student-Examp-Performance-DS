# 1. Setup Environment & Data Loading

```python
# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')

# Pengaturan agar visualisasi muncul dengan rapi dan bagus
%matplotlib inline
sns.set(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# 2. LOAD DATASET
df = pd.read_csv('StudentPerformanceFactors.csv')

# 3. QUICK CHECK
print("--- Dimensi Data ---")
print(f"Jumlah Baris: {df.shape[0]}")
print(f"Jumlah Kolom: {df.shape[1]}")

print("\n--- 5 Data Teratas ---")
display(df.head())

print("\n--- Jumlah Nilai Unik per Kolom ---")
print(df.nunique())

print("\n--- Cek Data Duplikat ---")
print(f"Jumlah baris duplikat: {df.duplicated().sum()}")
```

# 2. Handling Missing Values & Cleaning Data

```python
# 1. Cek missing values sebelum pembersihan
print("--- Missing Values Sebelum Pembersihan ---")
print(df.isnull().sum())

# 2. Mengisi data kosong dengan nilai yang paling sering muncul (Modus)
# Cocok untuk data kategorikal
df['Teacher_Quality'] = df['Teacher_Quality'].fillna(df['Teacher_Quality'].mode()[0])
df['Parental_Education_Level'] = df['Parental_Education_Level'].fillna(df['Parental_Education_Level'].mode()[0])
df['Distance_from_Home'] = df['Distance_from_Home'].fillna(df['Distance_from_Home'].mode()[0])

# 3. Verifikasi
print("\nJumlah data kosong setelah dibersihkan:", df.isnull().sum().sum())
```

## 2.1 Deteksi Anomali Data

Pada tahap ini kita memeriksa apakah terdapat nilai yang **melanggar batasan logis domain** (bukan sekadar outlier statistik).

**Konteks domain dataset ini:**
- `Exam_Score`: nilai ujian, rentang valid **0 – 100**
- `Hours_Studied`: jam belajar per minggu — tidak ada batas maksimum logis
- `Tutoring_Sessions`: jumlah sesi bimbingan — tidak ada batas maksimum logis

> **Kesimpulan Pemeriksaan:** Setelah dicek manual, **hanya ditemukan 1 anomali nyata** yaitu baris dengan `Exam_Score = 101` yang melebihi nilai maksimum yang mungkin (100). Nilai ini adalah *kesalahan entri data* dan akan dikoreksi pada sel berikutnya.

```python
# 1. Identifikasi baris anomali
anomaly_row = df[df['Exam_Score'] > 100]
print("Data anomali (Exam_Score > 100):")
print(anomaly_row[['Exam_Score']])

# 2. Memperbaiki anomali (Capping)
df.loc[df['Exam_Score'] > 100, 'Exam_Score'] = 100

# 3. Verifikasi hasil
print("\nSkor maksimal sekarang:", df['Exam_Score'].max())
```

# 3. Univariate Analysis (Mengenal Setiap Fitur)
Analisis distribusi untuk **semua fitur**, bukan hanya target variable.

```python
# 3a. Distribusi Fitur Numerik
numeric_cols = df.select_dtypes(include=[np.number]).columns.tolist()

fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    if i < len(axes):
        sns.histplot(df[col], kde=True, color='royalblue', ax=axes[i])
        axes[i].set_title(f'Distribusi {col}', fontsize=11)
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Frekuensi')

        # Tambahkan info skewness
        skew_val = df[col].skew()
        axes[i].annotate(f'Skewness: {skew_val:.2f}', xy=(0.65, 0.9),
                         xycoords='axes fraction', fontsize=9,
                         bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))

# Sembunyikan axes yang tidak terpakai
for j in range(i+1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Distribusi Semua Fitur Numerik', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
```

# 4. Deteksi Outlier
Menggunakan boxplot dan metode IQR untuk identifikasi outlier pada fitur numerik.

```python
# 4a. Boxplot untuk semua fitur numerik
fig, axes = plt.subplots(nrows=3, ncols=3, figsize=(16, 12))
axes = axes.flatten()

for i, col in enumerate(numeric_cols):
    if i < len(axes):
        sns.boxplot(y=df[col], ax=axes[i], color='lightcoral')
        axes[i].set_title(f'Boxplot {col}', fontsize=11)

for j in range(i+1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Deteksi Outlier - Boxplot Fitur Numerik', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# 4b. Hitung jumlah outlier per fitur (IQR method)
print("\n--- Jumlah Outlier per Fitur (IQR Method) ---")
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    if len(outliers) > 0:
        print(f"  {col}: {len(outliers)} outlier(s)")
    else:
        print(f"  {col}: Tidak ada outlier")
```

# 7. Feature Importance Analysis

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder
import pandas as pd

# 1. Copy dataframe agar data asli tidak rusak
df_encoded = df.copy()

# 2. Tentukan fitur ordinal vs nominal
ordinal_features = {
    'Parental_Involvement': ['Low', 'Medium', 'High'],
    'Access_to_Resources': ['Low', 'Medium', 'High'],
    'Motivation_Level': ['Low', 'Medium', 'High'],
    'Family_Income': ['Low', 'Medium', 'High'],
    'Teacher_Quality': ['Low', 'Medium', 'High'],
    'Peer_Influence': ['Negative', 'Neutral', 'Positive'],
    'Distance_from_Home': ['Near', 'Moderate', 'Far'],
    'Parental_Education_Level': ['High School', 'College', 'Postgraduate'],
}

nominal_features = ['Extracurricular_Activities', 'Internet_Access',
                    'Learning_Disabilities', 'School_Type', 'Gender']

# 3. Encode fitur ordinal dengan OrdinalEncoder (menjaga urutan)
for col, categories in ordinal_features.items():
    enc = OrdinalEncoder(categories=[categories])
    df_encoded[col] = enc.fit_transform(df_encoded[[col]]).astype(int)

# 4. Encode fitur nominal dengan OneHotEncoder (pd.get_dummies)
df_encoded = pd.get_dummies(df_encoded, columns=nominal_features, drop_first=True, dtype=int)

print("Shape setelah encoding:", df_encoded.shape)
print("\nKolom-kolom setelah encoding:")
print(df_encoded.columns.tolist())
display(df_encoded.head())
```

```python
# Tentukan urutan kolom sesuai urutan variabel di dataset asli
# dengan Exam_Score di posisi terakhir
ordered_cols = [
    'Hours_Studied', 'Attendance', 'Parental_Involvement', 'Access_to_Resources',
    'Extracurricular_Activities_Yes', 'Sleep_Hours', 'Previous_Scores',
    'Motivation_Level', 'Internet_Access_Yes', 'Tutoring_Sessions',
    'Family_Income', 'Teacher_Quality', 'School_Type_Public',
    'Peer_Influence', 'Physical_Activity', 'Learning_Disabilities_Yes',
    'Parental_Education_Level', 'Distance_from_Home', 'Gender_Male',
    'Exam_Score'  # Target variabel di posisi terakhir
]

# Hitung matriks korelasi dengan urutan kolom yang sudah ditentukan
corr_matrix = df_encoded[ordered_cols].corr()

# Buat heatmap korelasi
plt.figure(figsize=(18, 14))
sns.heatmap(
    corr_matrix,
    annot=True,
    fmt='.2f',
    cmap='coolwarm',
    center=0,
    linewidths=0.5,
    square=True,
    vmin=-1, vmax=1,
    cbar_kws={'shrink': 0.8, 'label': 'Korelasi'}
)
plt.title('Correlation Heatmap - Semua Variabel (Setelah Encoding)', fontsize=16, fontweight='bold', pad=20)
plt.xticks(rotation=45, ha='right', fontsize=9)
plt.yticks(fontsize=9)
plt.tight_layout()
plt.show()

# Tampilkan korelasi terhadap Exam_Score (diurutkan)
print("\n--- Korelasi terhadap Exam_Score (Semua Variabel) ---")
exam_corr = corr_matrix['Exam_Score'].drop('Exam_Score').sort_values(ascending=False)
print(exam_corr.to_string())
```

# 8. Data Preparation (Normalisasi & Split Data)
Pipeline yang konsisten: Encoding → Split → Scaling

```python
from sklearn.preprocessing import StandardScaler
import joblib

# 1. Tentukan Fitur (X) dan Target (y) - gunakan data yang sudah di-encode
X = df_encoded.drop('Exam_Score', axis=1)
y = df_encoded['Exam_Score']

# 2. Standardisasi / Normalisasi Seluruh Data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("--- Data Siap untuk Modelling ---")
print(f"Ukuran Fitur (X): {X_scaled.shape}")
print(f"Ukuran Target (y): {y.shape}")
print(f"\nRata-rata setelah scaling: {X_scaled.mean():.4f}")
print(f"Standar deviasi setelah scaling: {X_scaled.std():.4f}")

joblib.dump(scaler, 'scaler.pkl')
```

# 9. Modelling (Penerapan Metode Regresi)
Pada tahap ini, kita akan melatih dan mengevaluasi tiga model regresi yang berbeda untuk memprediksi `Exam_Score`. Semua model akan dilatih pada keseluruhan dataset yang telah diproses sebelumnya.


```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# ===== 1. LINEAR REGRESSION =====
# Latih model pada seluruh data
model_lr = LinearRegression()
model_lr.fit(X_scaled, y)

# Prediksi pada seluruh data
y_pred_lr = model_lr.predict(X_scaled)

# Evaluasi pada seluruh data
r2_lr = r2_score(y, y_pred_lr)
mae_lr = mean_absolute_error(y, y_pred_lr)
rmse_lr = np.sqrt(mean_squared_error(y, y_pred_lr))

print("=" * 50)
print("LINEAR REGRESSION (Evaluasi Umum)")
print("=" * 50)
print(f"R² (Umum): {r2_lr:.4f}")
print(f"MAE:       {mae_lr:.4f}")
print(f"RMSE:      {rmse_lr:.4f}")
```

```python
# ===== 2. RANDOM FOREST REGRESSION =====
from sklearn.ensemble import RandomForestRegressor

# Gunakan parameter terbaik dari tuning sebelumnya, atau parameter default yang solid
# Di sini kita latih ulang pada seluruh data
model_rf = RandomForestRegressor(
    n_estimators=100,
    max_depth=15,
    min_samples_leaf=2,
    min_samples_split=10,
    random_state=42,
    n_jobs=-1
)

model_rf.fit(X_scaled, y)

print("Model Random Forest telah dilatih pada seluruh dataset.")
```

```python
# Evaluasi Random Forest pada seluruh data
y_pred_rf = model_rf.predict(X_scaled)

r2_rf = r2_score(y, y_pred_rf)
mae_rf = mean_absolute_error(y, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y, y_pred_rf))

print("=" * 50)
print("RANDOM FOREST REGRESSION (Evaluasi Umum)")
print("=" * 50)
print(f"R² (Umum): {r2_rf:.4f}")
print(f"MAE:       {mae_rf:.4f}")
print(f"RMSE:      {rmse_rf:.4f}")
```

```python
import numpy as np
from sklearn.svm import SVR
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error

# ===== 3. SUPPORT VECTOR REGRESSION (SVR) =====
# Latih model pada seluruh data
model_svr = SVR(kernel='rbf', C=1.0, epsilon=0.1)
model_svr.fit(X_scaled, y)

# Prediksi pada seluruh data
y_pred_svr = model_svr.predict(X_scaled)

# Evaluasi pada seluruh data
r2_svr = r2_score(y, y_pred_svr)
mae_svr = mean_absolute_error(y, y_pred_svr)
rmse_svr = np.sqrt(mean_squared_error(y, y_pred_svr))

print("=" * 50)
print("SUPPORT VECTOR REGRESSION (Evaluasi Umum)")
print("=" * 50)
print(f"R² (Umum): {r2_svr:.4f}")
print(f"MAE:       {mae_svr:.4f}")
print(f"RMSE:      {rmse_svr:.4f}")
```
